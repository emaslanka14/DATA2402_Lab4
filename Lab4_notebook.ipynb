## Load CSV dataset
This loads the raw rows from the CSV file so we can see the input data before it is turned into objects

```python
teamsDict = {}
with open("f1_points.csv", "r") as f:
        for line in f: #Iterate through file until we find the header row
            if line.find(',') != -1:
                break
        for line in f:
            data = line.strip().split(",")
            try:
                driver = str(data[0])
                teamName = str(data[1])
                points = int(data[2])

                DriverObj = Driver(driver, points)
                if teamName not in teamsDict:
                    teamsDict[teamName] = Team(teamName)
                    teamsDict[teamName].add_driver(DriverObj)
                else:
                    teamsDict[teamName].add_driver(DriverObj)
            except Exception as e:
                print(f"Error: {e} in line: {line.strip()}")
                continue

        listOfTeams = list(teamsDict.values())
        sortedTeams = sorted(listOfTeams, reverse=True) #convert dict to list
```

## Special test case - Team with 3 drivers

**Input:** A new team with 3 drivers (50, 30 and 20 pts), plus two other teams with 150 and 60 total pts  
**Expected:** The team total is 100 and the teams sort from lowest to highest total using `<`  
**Actual:** The result is printed below

```python
special = Team("TEST RACING")
special.add_driver(Driver("Driver A", 50))
special.add_driver(Driver("Driver B", 30))
special.add_driver(Driver("Driver C", 20))

high = Team("HIGH TEAM")
high.add_driver(Driver("Driver D", 150))

low = Team("LOW TEAM")
low.add_driver(Driver("Driver E", 60))

print(special)
print("Total points:", special.get_total_points())

print("\nSorted (lowest to highest):")
for team in sorted([high, special, low]):
    print(team.name,":", team.get_total_points())
```
