## Task 1 Load CSV dataset
This loads the raw rows from the CSV file so we can see the input data before it is turned into objects

```python
teams = {}
with open("f1_points.csv", "r") as f:

    reader = csv.reader(f)  

    next(reader)  # Skip the header row
    for line in reader:
        try:
            driver = str(line[0])
            teamName = str(line[1])
            points = int(line[2])

            DriverObj = Driver(driver, points)
            if teamName not in teams:
                teams[teamName] = Team(teamName)
            
            teams[teamName].add_driver(DriverObj)
        except Exception as e:
            print(f"Error: {e} in line: {line.strip()}")
            continue

    print(f"Number of Teams = {len(teams)}")

    listOfTeams = list(teams.values())
    sortedTeams = sorted(listOfTeams, reverse=True) #convert dict to list
```
